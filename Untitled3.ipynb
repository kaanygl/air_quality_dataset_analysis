```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_brown = pd.read_csv('processed_brown.csv')
df_gold = pd.read_csv('processed_gold.csv')
df_bsilver = pd.read_csv('processed_silver.csv')
```

```python
df = pd.concat([df_brown, df_gold, df_bsilver])
```

```python
df.isna().sum()
```

```python
for i in df.columns:
  if df[i].isnull().any():
    df[i].fillna(df[i].mean(), inplace=True)
```

```python
df.isna().sum()
```

```python
df['ts_insertion'] = pd.to_datetime(df['ts_insertion'])
df['node_id']=df['node_id'].astype('category').cat.codes
corr = df.corr()
plt.figure(figsize=(20, 10))
sns.heatmap(corr, cmap="Reds", annot=True)
plt.show()
```
