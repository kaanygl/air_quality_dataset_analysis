```python
df = pd.read_csv("C://complete_measure_node_airport_gold.csv", delimiter=';')

df
```

```python
#Turning null values into 0's
columns_to_fill = [
    'sps30_pm05_count', 'sps30_pm1_count', 'sps30_pm25_count', 'sps30_pm4_count',
    'sps30_pm10_count', 'sps30_pm_typ', 'sps30_pm25_ug', 'sps30_pm4_ug', 'sps30_pm10_ug'
]

df[columns_to_fill] = df[columns_to_fill].fillna(0)

print(df.isnull().sum())
```

```python
cols_to_drop=['id_measure','node_id']

df = df.drop(columns=df.filter(items=cols_to_drop).columns)

df
```

```python
numeric_cols = df.select_dtypes(include=['float64', 'int64'])

Q1 = numeric_cols.quantile(0.25)
Q3 = numeric_cols.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 5 * IQR
upper_bound = Q3 + 5 * IQR

outliers = (numeric_cols < lower_bound) | (numeric_cols > upper_bound)

outliers_count = outliers.sum()
print(outliers_count)
```

```python
# Replace outliers for ads1115_value
df['ads1115_value'] = np.where(
    (df['ads1115_value'] < 0) | (df['ads1115_value'] > 5000),  # Custom thresholds for ads1115_value
    df['ads1115_value'].median(),
    df['ads1115_value']
)

# Replace outliers for ads1115_voltage
df['ads1115_voltage'] = np.where(
    (df['ads1115_voltage'] < 0.1) | (df['ads1115_voltage'] > 0.35),  # Assuming voltage is between 0 and 5
    df['ads1115_voltage'].median(),
    df['ads1115_voltage']
)

# Replace outliers for scd30_co2
df['scd30_co2'] = np.where(
    (df['scd30_co2'] < 300) | (df['scd30_co2'] > 1000),  # Maximum threshold for CO2
    df['scd30_co2'].median(),
    df['scd30_co2']
)

# Replace outliers for scd30_temp
df['scd30_temp'] = np.where(
    (df['scd30_temp'] < 15) | (df['scd30_temp'] > 40),  # Custom thresholds for temperature
    df['scd30_temp'].median(),
    df['scd30_temp']
)

# Replace outliers for scd30_hum (no outliers detected, no replacement necessary)
# df['scd30_hum'] remains unchanged.

# Replace outliers for sps30_pm05_count
df['sps30_pm05_count'] = np.where(
    (df['sps30_pm05_count'] < 10) | (df['sps30_pm05_count'] > 60),  # Assuming counts are between 0 and 100
    df['sps30_pm05_count'].median(),
    df['sps30_pm05_count']
)

# Replace outliers for sps30_pm1_count
df['sps30_pm1_count'] = np.where(
    (df['sps30_pm1_count'] < 0) | (df['sps30_pm1_count'] > 5),
    df['sps30_pm1_count'].median(),
    df['sps30_pm1_count']
)

# Replace outliers for sps30_pm1_ug
df['sps30_pm1_ug'] = np.where(
    (df['sps30_pm1_ug'] < 1) | (df['sps30_pm1_ug'] > 3),
    df['sps30_pm1_ug'].median(),
    df['sps30_pm1_ug']
)

# Replace outliers for sps30_pm25_count
df['sps30_pm25_count'] = np.where(
    (df['sps30_pm25_count'] < 0) | (df['sps30_pm25_count'] > 3),
    df['sps30_pm25_count'].median(),
    df['sps30_pm25_count']
)

# Replace outliers for sps30_pm25_ug
df['sps30_pm25_ug'] = np.where(
    (df['sps30_pm25_ug'] < 0) | (df['sps30_pm25_ug'] > 3),
    df['sps30_pm25_ug'].median(),
    df['sps30_pm25_ug']
)

# Replace outliers for sps30_pm4_count
df['sps30_pm4_count'] = np.where(
    (df['sps30_pm4_count'] < 0) | (df['sps30_pm4_count'] > 3),
    df['sps30_pm4_count'].median(),
    df['sps30_pm4_count']
)

# Replace outliers for sps30_pm4_ug
df['sps30_pm4_ug'] = np.where(
    (df['sps30_pm4_ug'] < 0) | (df['sps30_pm4_ug'] > 3),
    df['sps30_pm4_ug'].median(),
    df['sps30_pm4_ug']
)

# Replace outliers for sps30_pm10_count
df['sps30_pm10_count'] = np.where(
    (df['sps30_pm10_count'] < 0) | (df['sps30_pm10_count'] > 3),
    df['sps30_pm10_count'].median(),
    df['sps30_pm10_count']
)

# Replace outliers for sps30_pm10_ug
df['sps30_pm10_ug'] = np.where(
    (df['sps30_pm10_ug'] < 0) | (df['sps30_pm10_ug'] > 3),
    df['sps30_pm10_ug'].median(),
    df['sps30_pm10_ug']
)

# Replace outliers for sps30_pm_typ
df['sps30_pm_typ'] = np.where(
    (df['sps30_pm_typ'] < 0) | (df['sps30_pm_typ'] > 3),  # Assuming it's a normalized column
    df['sps30_pm_typ'].median(),
    df['sps30_pm_typ']
)

print("Outliers handled using domain-specific thresholds.")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt

scaler = MinMaxScaler()

columns = [
    'ads1115_value', 'ads1115_voltage', 'scd30_co2', 'scd30_temp',
    'sps30_pm05_count', 'sps30_pm1_count', 'sps30_pm1_ug',
    'sps30_pm25_count', 'sps30_pm25_ug', 'sps30_pm4_count',
    'sps30_pm4_ug', 'sps30_pm10_count', 'sps30_pm10_ug', 'sps30_pm_typ'
]

df_normalized = df.copy()
df_normalized[columns] = scaler.fit_transform(df[columns])

n_cols = 3 
n_rows = int(np.ceil(len(columns) / n_cols)) 
fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 3))

axes = axes.flatten()

for i, col in enumerate(columns):
    axes[i].hist(df_normalized[col], bins=300, alpha=0.7, color='green')
    axes[i].set_title(f'{col} (Normalized)', fontsize=10)
    axes[i].set_xlabel(col, fontsize=8)
    axes[i].set_ylabel('Frequency', fontsize=8)

for j in range(len(columns), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
